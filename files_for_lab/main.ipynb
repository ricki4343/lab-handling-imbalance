```python
churnData['TotalCharges'] = pd.to_numeric(churnData['TotalCharges'], errors='coerce')
```

```python
churnData['TotalCharges'].fillna(0, inplace=True)
```

#

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

X = churnData[['tenure', 'SeniorCitizen', 'MonthlyCharges', 'TotalCharges']]
y = (churnData['Churn'] == 'Yes').astype(int)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=0)

model = LogisticRegression(random_state=0)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print(classification_report(y_test, y_pred, target_names=['No Churn', 'Churn']))
```

#

```python
# separate two classes
from sklearn.utils import resample

churn_majority = churnData[churnData['Churn'] == 'No']
churn_minority = churnData[churnData['Churn'] == 'Yes']
```

```python
# Upsampling

churn_minority_up = resample(churn_minority,
                                    replace=True,
                                    n_samples=len(churn_majority),  # match majority count
                                    random_state=0)

churn_upsampled = pd.concat([churn_majority, churn_minority_up])
print(churn_upsampled['Churn'].value_counts())
```

```python
# Downsampling

churn_majority_down = resample(churn_majority,
                                      replace=False,                  # sample without replacement
                                      n_samples=len(churn_minority),  # match minority count
                                      random_state=0)

churn_downsampled = pd.concat([churn_majority_down, churn_minority])
print(churn_downsampled['Churn'].value_counts())
```

#

```python
# Upsample

Xu = churn_upsampled[['tenure', 'SeniorCitizen', 'MonthlyCharges', 'TotalCharges']]
yu = (churn_upsampled['Churn'] == 'Yes').astype(int)

scaler = StandardScaler()
Xu_scaled = scaler.fit_transform(Xu)

Xu_train, Xu_test, yu_train, yu_test = train_test_split(Xu_scaled, yu, test_size=0.2, random_state=0)

model = LogisticRegression(random_state=0)
model.fit(Xu_train, yu_train)

yu_pred = model.predict(Xu_test)
```

```python
# Downsample

Xd = churn_downsampled[['tenure', 'SeniorCitizen', 'MonthlyCharges', 'TotalCharges']]
yd = (churn_downsampled['Churn'] == 'Yes').astype(int)

scaler = StandardScaler()
Xd_scaled = scaler.fit_transform(Xd)

Xd_train, Xd_test, yd_train, yd_test = train_test_split(Xd_scaled, yd, test_size=0.2, random_state=0)

model = LogisticRegression(random_state=0)
model.fit(Xd_train, yd_train)

yd_pred = model.predict(Xd_test)
```

```python
print("Upsample")
print(f"Accuracy: {accuracy_score(yu_test, yu_pred):}")
print(classification_report(yu_test, yu_pred, target_names=['No Churn', 'Churn']))

print("Downsample")
print(f"Accuracy: {accuracy_score(yd_test, yd_pred):}")
print(classification_report(yd_test, yd_pred, target_names=['No Churn', 'Churn']))
```

both accuracy are lower, but churn recall are much much better
